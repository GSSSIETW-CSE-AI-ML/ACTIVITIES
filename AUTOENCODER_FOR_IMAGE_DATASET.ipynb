```python
# ============================================================
# EXPERIMENT 4
# Build and Demonstrate an Autoencoder Network Using
# Neural Layers for Data Compression on an Image Dataset
# Dataset: MNIST
# ============================================================


# ------------------------------------------------------------
# STEP 1: Import Required Libraries
# ------------------------------------------------------------

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers
from tensorflow.keras import models


# ------------------------------------------------------------
# STEP 2: Load MNIST Dataset
# ------------------------------------------------------------

(x_train, _), (x_test, _) = tf.keras.datasets.mnist.load_data()

print("Training data shape:", x_train.shape)
print("Testing data shape :", x_test.shape)


# ------------------------------------------------------------
# STEP 3: Display Sample Images
# ------------------------------------------------------------

plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.axis("off")

plt.suptitle("Sample MNIST Images")
plt.show()


# ------------------------------------------------------------
# STEP 4: Normalize Pixel Values
# ------------------------------------------------------------

# Pixel values are originally between 0 and 255.
# Convert them to the range 0 to 1.

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Minimum pixel value:", x_train.min())
print("Maximum pixel value:", x_train.max())


# ------------------------------------------------------------
# STEP 5: Flatten the Images
# ------------------------------------------------------------

# MNIST image size = 28 x 28
# 28 x 28 = 784

x_train = x_train.reshape((len(x_train), 784))
x_test = x_test.reshape((len(x_test), 784))

print("After flattening:")
print("Training data shape:", x_train.shape)
print("Testing data shape :", x_test.shape)


# ------------------------------------------------------------
# STEP 6: Define Input Layer
# ------------------------------------------------------------

input_img = layers.Input(shape=(784,), name="Input")


# ------------------------------------------------------------
# STEP 7: Build the Encoder
# ------------------------------------------------------------

encoded = layers.Dense(
    128,
    activation="relu",
    name="Encoder_128"
)(input_img)

encoded = layers.Dense(
    64,
    activation="relu",
    name="Encoder_64"
)(encoded)

# Latent Space / Bottleneck
encoded = layers.Dense(
    32,
    activation="relu",
    name="Latent_Space"
)(encoded)


# ------------------------------------------------------------
# STEP 8: Build the Decoder
# ------------------------------------------------------------

decoded = layers.Dense(
    64,
    activation="relu",
    name="Decoder_64"
)(encoded)

decoded = layers.Dense(
    128,
    activation="relu",
    name="Decoder_128"
)(decoded)

decoded = layers.Dense(
    784,
    activation="sigmoid",
    name="Output"
)(decoded)


# ------------------------------------------------------------
# STEP 9: Create the Complete Autoencoder
# ------------------------------------------------------------

autoencoder = models.Model(
    input_img,
    decoded,
    name="MNIST_Autoencoder"
)


# ------------------------------------------------------------
# STEP 10: Display Model Architecture
# ------------------------------------------------------------

autoencoder.summary()


# ------------------------------------------------------------
# STEP 11: Compile the Autoencoder
# ------------------------------------------------------------

autoencoder.compile(
    optimizer="adam",
    loss="binary_crossentropy"
)


# ------------------------------------------------------------
# STEP 12: Train the Autoencoder
# ------------------------------------------------------------

history = autoencoder.fit(
    x_train,
    x_train,              # Input = Target
    epochs=20,
    batch_size=256,
    shuffle=True,
    validation_data=(x_test, x_test)
)


# ------------------------------------------------------------
# STEP 13: Plot Training and Validation Loss
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Autoencoder Training Loss")

plt.legend()
plt.grid()

plt.show()


# ------------------------------------------------------------
# STEP 14: Create the Encoder Model Separately
# ------------------------------------------------------------

encoder = models.Model(
    input_img,
    encoded,
    name="Encoder"
)


# ------------------------------------------------------------
# STEP 15: Generate Compressed Representations
# ------------------------------------------------------------

encoded_imgs = encoder.predict(x_test)

print("\nOriginal image representation:")
print("784 values")

print("\nCompressed representation:")
print(encoded_imgs.shape)


# ------------------------------------------------------------
# STEP 16: Display One Latent Representation
# ------------------------------------------------------------

print("\nLatent representation of first test image:")
print(encoded_imgs[0])

print("\nNumber of latent values:",
      len(encoded_imgs[0]))


# ------------------------------------------------------------
# STEP 17: Reconstruct Test Images
# ------------------------------------------------------------

decoded_imgs = autoencoder.predict(x_test)


# ------------------------------------------------------------
# STEP 18: Display Original vs Reconstructed Images
# ------------------------------------------------------------

n = 10

plt.figure(figsize=(20, 4))

for i in range(n):

    # Original Image
    ax = plt.subplot(2, n, i + 1)

    plt.imshow(
        x_test[i].reshape(28, 28),
        cmap="gray"
    )

    plt.title("Original")
    plt.axis("off")


    # Reconstructed Image
    ax = plt.subplot(2, n, i + 1 + n)

    plt.imshow(
        decoded_imgs[i].reshape(28, 28),
        cmap="gray"
    )

    plt.title("Reconstructed")
    plt.axis("off")


plt.suptitle(
    "Original Images vs Reconstructed Images"
)

plt.show()


# ------------------------------------------------------------
# STEP 19: Calculate Compression Ratio
# ------------------------------------------------------------

original_dimension = 784
latent_dimension = 32

compression_ratio = (
    original_dimension / latent_dimension
)

print("\nOriginal dimension :", original_dimension)
print("Latent dimension   :", latent_dimension)

print(
    "Approximate dimensionality reduction ratio:",
    compression_ratio
)


# ------------------------------------------------------------
# STEP 20: Calculate Reconstruction Error
# ------------------------------------------------------------

mse = np.mean(
    np.square(x_test - decoded_imgs)
)

print("\nMean Squared Reconstruction Error:",
      mse)


# ============================================================
# EXPERIMENT COMPLETE
# ============================================================
```
